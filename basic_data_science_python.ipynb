String *slices* can be accessed by referencing positions separated by a colon `:`.
The characters are selected up to, but not including, the second position.

```python
# "string"[0:2] are the first two characters
"STRING"[0:2]
```

```python
# "string"[1:3] are the second character up to, but not including, the fourth character (i.e., characters 1 and 2)
"STRING"[1:3]
```

```python
# with the first position omitted, "string"[:2] are the first two characters
"STRING"[:2]
```

```python
# with the second position omitted, "string"[3:] are the last three characters
"STRING"[3:]
```

```python
# slices can include negative positions, "string"[-3:-1] are the third to last up to, but not including, the last character
"STRING"[-3:-1]
```

```python
# slices with a negative position omitting the second position, "string"[-4:] are the four characters from position -4 (included) to the last character
"STRING"[-4:]
```

```python
# slices with a negative position omitting the first position, "string"[:-3] are the first three characters, up to and but not including, position -3
"STRING"[:-3]
```

## Booleans

Booleans are the result of evaluating conditions or comparisons. They have two values: True and False. They can be considered Nominal data or a label of two separate states. - \>= (greater than or equal)
- < (less than)
- <= (less than or equal)

```python
# A true condition
4 > 3
```

- list[0] is the first element
- list[1:3] are the first *up to* the third element (i.e., elements 1 and 2)
- list[-1] is the last element

## Sets
Sets are an unordered group of values with no duplicates. Sets have rich set of operations as defined by [set theory](https://realpython.com/python-sets/), of which a useful set are implemented in Python.

```python
# set without duplicates
{'apple', 'orange', 'apple', 'pear', 'orange', 'banana'}
```

```python
# count the number of values in the set
len({'apple', 'banana', 'orange', 'pear'})
```

```python
# detecting if a value is in a set
'orange' in {'apple', 'banana', 'orange', 'pear'}
```

```python
# detecting if a value is not in a set
'grape' in {'apple', 'banana', 'orange', 'pear'}
```

```python
# set intersection: fruit in both sets
{'apple', 'banana', 'orange', 'pear'} & {'grape', 'banana', 'peach', 'orange'}
```

```python
# set union: fruit in either set
{'apple', 'banana', 'orange', 'pear'} | {'grape', 'banana', 'peach', 'orange'}
```

```python
# set difference: fruit in one set but not in another
{'apple', 'banana', 'orange', 'pear'} - {'grape', 'banana', 'peach', 'orange'}
```

```python
# fruit in one or the othere set but not both
{'apple', 'banana', 'orange', 'pear'} ^ {'grape', 'banana', 'peach', 'orange'}
```

```python
# is one set a subset of another
{'banana', 'pear'} <= {'apple', 'banana', 'orange', 'pear'}
```

```python
# is one set a superset of another
{'apple', 'banana', 'orange', 'pear'} >= {'banana', 'pear'}
```

operations that modify a set

```python
# add a value to set
basket = {'apple', 'banana', 'orange', 'pear'}
basket.add('orange')
basket
```

```python
# remove a value from a set
basket = {'apple', 'banana', 'orange', 'pear'}
basket.remove('pear')
basket
```

```python
# remove all values from a set
basket = {'apple', 'banana', 'orange', 'pear'}
basket.clear()
basket
```

## Dictionaries
Dictionaries are a group where you look up a value with another value, a key, like looking up a word definition in a real dictionary. Dictionaries are formed with enclosing pairs of values of any type between braces {}. Dictionary values are accessed by key using square brackets.


```python
# a sample dictionary
{'leaf': 'green', 'sky': 'blue', 'apple': 'red'}
```

```python
# select a dictionary value
{'leaf': 'green', 'sky': 'blue', 'apple': 'red'}['leaf']
```

```python
# count the dictionary pairs 
len({'leaf': 'green', 'sky': 'blue', 'apple': 'red'})
```

```python
#  detect if a key is in a dictionary
'leaf' in {'leaf': 'green', 'sky': 'blue', 'apple': 'red'}
```

```python
#  detect if a key is not in a dictionary
'cloud' in {'leaf': 'green', 'sky': 'blue', 'apple': 'red'}
```

```python
# list all the keys in a dictionary
{'leaf': 'green', 'sky': 'blue', 'apple': 'red'}.keys()
```

```python
# list all the values in a dictionary
{'leaf': 'green', 'sky': 'blue', 'apple': 'red'}.values()
```

operations that modify a dictionary

```python
dictionary1 = {'leaf': 'green', 'sky': 'blue', 'apple': 'red'}
# count the pairs in a dictionary
print("3. ", len(dictionary1))
# change a values in a dictionary
dictionary1["apple"] = "yellow"
print("4. ", dictionary1)
# adding values to a dictionary
dictionary1["earth"] = "brown"
print("5. ", dictionary1)
# removing a value from a dictionary
del dictionary1["sky"]
print("6. ", dictionary1)

# list all the values in a dictionary
print("10. ", dictionary1.values())
# combine two dictionaries
dictionary2 = {"apple": "yellow", "sunset": "orange"}
dictionary1.update(dictionary2)
print("11. ", dictionary1)
# remove all pairs in a dictionarie
dictionary1.clear()
print("12. ", dictionary1)
```
